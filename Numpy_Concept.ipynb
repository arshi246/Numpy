# 8. Vectorized Operations

This is one of the most important features of NumPy.

NumPy can perform one operation on all values at the same time.

````python
numbers = np.array([10, 20, 30, 40])
````

Add:

````python
print(numbers + 5)
````

Subtract:

````python
print(numbers - 5)
````

Multiply:

````python
print(numbers * 5)
````

Divide:

````python
print(numbers / 5)
````

This is called a **vectorized operation**.

We do not need to use a loop for every value.

Agar har operation ko numbers mein store karwao
Har line ke baad numbers update hota rahega:
Toh end mein numbers ki value last operation ka result hogi.

````python
numbers = np.array([10, 20, 30, 40])

numbers = numbers + 5
numbers = numbers - 2
numbers = numbers * 3
numbers = numbers / 2

print(numbers)
````

Specific value ko change karna ho tw :

````python
numbers = np.array([10, 20, 30, 40])
numbers[1] = 100
print(numbers)
````

Specific value me calculation karni ho tw :


Calculate + specific index mein store → original array updated
Yani = sabse important cheez hai. = ke left side par jo jagah hoti hai, result wahan store hota hai.

````python
numbers[2] = numbers[2] + 5
print(numbers)
````

or agar ham variable me store nh karwayengay tw wo original cahnge nahi karega

````python
numbers = np.array([10, 20, 30, 40])

numbers[1] * 2

print(numbers)
````

# 14. Reshape

Important rule:

The total number of values must stay the same.

```
2 × 3 = 6
```

So the reshape is possible.

`reshape()` changes the structure of an array.

Example:

````python
numbers = np.array([1, 2, 3, 4, 5, 6])
print(numbers)
````

Now convert it into 2 rows and 3 columns:

````python
new_array = numbers.reshape(2,3)
print(new_array )
````

# Class Practice: Student Marks Analyzer

Now we will use different NumPy concepts in one small program.

## Problem

We have marks of different students.

We want to find:

- Average marks
- Highest marks
- Lowest marks
- Total passed students
- Total failed students

````python
marks = np.array([45, 78, 90, 32, 67, 88, 55, 40, 95, 28])

print("Student Marks:")
print(marks)

# Average Marks
print(f"\nAverage Marks : {np.mean(marks)}")

# Highest marks
print(f"Highest Marks : {np.max(marks)}")

# Lowest marks
print(f"Lowest Marks : {np.min(marks)}")

passed = marks[marks>=50]
failed = marks[marks<50]


print(f"\nPassed Students Marks : {passed}")
print(f"\nFailed Students Marks : {failed}")

# Total passed students
print(f"Total passed students : {len(passed)}")


# Total failed students
print(f"Total Failed students : {len(failed)}")
````
